```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat.avif")
cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
# Screenshot example: 463 => height, 703 => width
# 3 => color channels (BGR) => Blue, Green, Red
print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (300, 300))
cv2_imshow(resized)
```

```python
cropped = gray[100:300, 150:350]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
# Create the flipped image used in the save example
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped_cat.jpg", flipped)
```

```python
# Save the crop before reading it back in the next cell
cv2.imwrite("/content/cropped_cat.jpg", cropped)
```

```python
cropped_cat = cv2.imread("/content/cropped_cat.jpg")
cv2_imshow(cropped_cat)
```

Open Cv Video Codes

```python
print("Hello World!")

import cv2

url = "rtsp://192.168.1.164:8001/video"  # stream_path
cap = cv2.VideoCapture(url)

while True:
    success, frame = cap.read()

    if not success:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Original", frame)
    cv2.imshow("Gray", gray)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

#===================================================

# import cv2

# url = "rtsp://192.168.1.164:8001/video"  # stream_path
# cap = cv2.VideoCapture(url)

# while True:
#     success, frame = cap.read()  # True, False | Boolean

#     if not success:
#         break

#     gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
#     blurred = cv2.GaussianBlur(frame, (15, 15), 0)
#     edges = cv2.Canny(frame, 100, 200)
#     edges2 = cv2.Canny(frame, 50, 250)

#     flipped = cv2.flip(frame, 1)

#     cv2.imshow("Original", frame)
#     # cv2.imshow("Gray", gray)
#     # cv2.imshow("Blurred", blurred) # Noise
#     # cv2.imshow("Edges", edges)
#     # cv2.imshow("Edges2", edges2)

#     cv2.imshow("Flipped", flipped)


#     if cv2.waitKey(1) & 0xFF == ord('q'):
#         break

# cap.release()
# cv2.destroyAllWindows()
```
